### 🚀 **Quick Start Guide**
1. **Enable GPU:** Go to `Runtime > Change runtime type` in the top menu and select **T4 GPU**.
2. **Add Token (Optional):** Click the 🔑 icon on the left to add a secret named `HF_TOKEN` with your Hugging Face token to speed up downloads.
3. **Save to Drive (Optional):** Tick `save_models_to_drive` in the cell below so later sessions skip the ~5 GB model download.
4. **Run the Cell Below:** Click the play button. The first run takes about 10 minutes (installing and downloading); later runs are much faster.
5. A public **`xxxx.gradio.live`** link will appear at the bottom. Click it to open your UI!

> ⚠️ Only lip-sync videos of people who have given their consent. Anyone with the public link can use your running app.

```python
# @title 🚀 Install and Run MuseTalk Studio
import os

save_models_to_drive = False  # @param {type:"boolean"}

repo_url = "https://github.com/MubeenAmjad205/MuseTalk.git"
repo_dir = "/content/MuseTalk"

if not os.path.exists(repo_dir):
    print("1. Cloning Repository...")
    !git clone --depth 1 {repo_url} {repo_dir}
else:
    print("1. Updating Repository to the latest version...")
    !git -C {repo_dir} fetch --depth 1 origin main -q && git -C {repo_dir} reset --hard origin/main -q

print("\n2. Entering project directory...")
os.chdir(repo_dir)

print("\n3. Checking for Hugging Face Token in Colab Secrets...")
try:
    from google.colab import userdata
    hf_token = userdata.get('HF_TOKEN')
    if hf_token:
        os.environ["HF_TOKEN"] = hf_token
        print("✅ HF_TOKEN found and loaded successfully! Downloads will be faster.")
    else:
        print("⚠️ HF_TOKEN is empty in Colab Secrets. Proceeding without it.")
except Exception:
    print("⚠️ HF_TOKEN not found in Colab Secrets. Proceeding without it (downloads may be slower).")

if save_models_to_drive:
    from google.colab import drive
    drive.mount("/content/drive")
    os.environ["MUSETALK_DRIVE_DIR"] = "/content/drive/MyDrive/MuseTalk_models"

print("\n4. Setting up and launching MuseTalk Studio...")
!python scripts/run_colab.py
```
